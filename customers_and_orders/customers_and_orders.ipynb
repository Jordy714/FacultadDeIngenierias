**Question 1:** How many customers are in the file?
(as help, we have added some comments and starter code to help you structure the solution)

```python
from os.path import exists
import csv
from collections import Counter

# check if file can be found and open the file
datafile = "./sample_data/customers.csv"
if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

# customers file is in the sample_data folder and can be opened.
# Using csv reader is one way of dealing with the file
# but it's up to you how you want to proceed
with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')
  header = next(csvreader)
  customers = list(csvreader)
  # loop through the rows and count etc. THE REST IS UP TO YOU!

# Eliminar filas completamente duplicadas
customers = list(dict.fromkeys(tuple(row) for row in customers))

# Convertir las filas en diccionarios para trabajar por nombre de columna
customers = [dict(zip(header, row)) for row in customers]

print("Total number of customers:", len(customers))
print("All done!")
```

**Question 2:** In how many different states do the customers live in?

```python

# open file and init vars
# evaluate the contents of the file
states = Counter(
    row["State"].strip()
    for row in customers
    if row["State"].strip()
)

print("Different states:", len(states))
```

**Question 3** What is the state with most customers?

```python
# evaluate the contents of the file
states_count = Counter(
    row["State"].strip()
    for row in customers
    if row["State"].strip()
)

print("State with most customers:", states_count.most_common(1)[0])
```

**Question 4** What is the state with the least customers?

```python
# evaluate the contents of the file
states_count = Counter(
    row["State"].strip()
    for row in customers
    if row["State"].strip()
)

print("State with least customers:",
      min(states_count.items(), key=lambda item: item[1]))
```

**Question 5:** What is the most common last name?

```python
# evaluate the contents of the file
last_names = Counter(
    row["LastName"].strip()
    for row in customers
    if row["LastName"].strip()
)

print("Most common last name:", last_names.most_common(1)[0])
```

**Question #1:** How many unique orders are in the orders.csv file?

**Question #2:** What is the average number of items per order (rounded to two decimal places)?

**Question #3:** What is the highest number of items per order?

**Question #4:** What is the number of orders placed in October 2021?

**Question #5:** Which customer spent the most amount of money in 2021?

**Question #6:** Historically, what is the best month for sales?

Once you get your answers, remember to go back to the course and introduce them in the multiple choice quiz

```python
from datetime import datetime

# open and evaluate orders.csv
datafileOrders = "./sample_data/orders.csv"

# check if file can be found and open the file
if not exists(datafileOrders):
  raise SystemExit("You should run the first code cell and download the dataset files!")

with open(datafileOrders, "r") as fl:
    csvreaderOrders = csv.DictReader(fl)
    orders = list(csvreaderOrders)

orders = list({
    tuple(row.items()): row
    for row in orders
}.values())
#question 1
unique_orders = {
    row["OrderID"].strip()
    for row in orders
    if row["OrderID"].strip()
}
print("Unique orders:", len(unique_orders))

#question 2
items_per_order = Counter(
    row["OrderID"].strip()
    for row in orders
    if row["OrderID"].strip()
)
average_items = sum(items_per_order.values()) / len(items_per_order)
print("Average items per order:", round(average_items, 2))

#question 3
print("Highest number of items per order:", max(items_per_order.values()))

#question 4
empty_dates = sum(
    1 for row in orders
    if not row["Date"].strip()
)

print("Rows with empty dates:", empty_dates)

october_orders = set()

for row in orders:
    order_id = row["OrderID"].strip()
    date_text = row["Date"].strip()

    if not order_id or not date_text:
        continue

    try:
        date = datetime.strptime(
            date_text[:19], "%Y-%m-%d %H:%M:%S"
        )
    except ValueError:
        continue

    if date.year == 2021 and date.month == 10:
        october_orders.add(order_id)

print("Orders in October 2021:", len(october_orders))

#question 5
spending = {}
for row in orders:
  date_text = row["Date"].strip()

  if not date_text:
      continue

  try:
      date = datetime.strptime(
          date_text[:19], "%Y-%m-%d %H:%M:%S"
      )
  except ValueError:
      continue
  if date.year == 2021:
      customer_id = row["CustomerID"].strip()
      order_id = row["OrderID"].strip()

      # Evitar sumar varias veces el total del mismo pedido
      spending.setdefault(customer_id, {})
      spending[customer_id][order_id] = float(row["OrderTotal"])

customer_spending = {
    customer_id: sum(order_totals.values())
    for customer_id, order_totals in spending.items()
}

print("Customer who spent the most:",
      max(customer_spending.items(), key=lambda item: item[1]))

#question 6
sales_by_month = {}

for row in orders:
  date_text = row["Date"].strip()

  if not date_text:
      continue

  try:
      date = datetime.strptime(
          date_text[:19], "%Y-%m-%d %H:%M:%S"
      )
  except ValueError:
      continue
  order_id = row["OrderID"].strip()
  month = date.month

  sales_by_month.setdefault(month, {})
  sales_by_month[month][order_id] = float(row["OrderTotal"])

monthly_sales = {
    month: sum(order_totals.values())
    for month, order_totals in sales_by_month.items()
}

best_month = max(monthly_sales.items(), key=lambda item: item[1])

print("Best month for sales:",
      datetime(2000, best_month[0], 1).strftime("%B"))
print("Total sales:", round(best_month[1], 2))


```
